```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

target="PlacementStatus"; drops=["PlacementStatus","Salary Package","StudentID","IsAnomaly"]
X=df.drop(columns=drops,errors="ignore"); y=df[target].astype(int)
num=X.select_dtypes(include=np.number).columns.tolist(); cat=X.select_dtypes(include="object").columns.tolist()
pre=ColumnTransformer([("num",SimpleImputer(strategy="median"),num),
("cat",Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("onehot",OneHotEncoder(handle_unknown="ignore"))]),cat)])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
for name,clf in [("Decision Tree",DecisionTreeClassifier(max_depth=5,random_state=42)),
                 ("Random Forest",RandomForestClassifier(n_estimators=100,max_depth=12,random_state=42,n_jobs=2))]:
    pipe=Pipeline([("preprocessor",pre),("model",clf)])
    pipe.fit(Xtr,ytr); pred=pipe.predict(Xte)
    print(name)
    print("Accuracy:",accuracy_score(yte,pred))
    print("Precision:",precision_score(yte,pred))
    print("Recall:",recall_score(yte,pred))
    print("F1 Score:",f1_score(yte,pred))
    print("-"*40)
```

```python
# Train both models once more for a direct comparison
dt=Pipeline([("preprocessor",pre),("model",DecisionTreeClassifier(max_depth=5,random_state=42))])
rf=Pipeline([("preprocessor",pre),("model",RandomForestClassifier(n_estimators=100,max_depth=12,random_state=42,n_jobs=2))])
dt.fit(Xtr,ytr); rf.fit(Xtr,ytr)
results=pd.DataFrame({
"Model":["Decision Tree","Random Forest"],
"Accuracy":[accuracy_score(yte,dt.predict(Xte)),accuracy_score(yte,rf.predict(Xte))]
})
display(results)
plt.bar(results["Model"],results["Accuracy"])
plt.ylim(0,1); plt.ylabel("Accuracy"); plt.title("Decision Tree vs Random Forest")
plt.show()
```
