```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.ensemble import IsolationForest

features=["CGPA","AttendancePercent","Internships","Projects","Certifications","AptitudeTestScore","CodingTestScore","SoftSkillsRating","MockInterviewScore"]
X=df[features].fillna(df[features].median())
X_scaled=StandardScaler().fit_transform(X)

# DBSCAN is run on a representative sample for speed on 50,000 rows.
rng=np.random.RandomState(42)
idx=rng.choice(len(X_scaled),size=min(10000,len(X_scaled)),replace=False)
db=DBSCAN(eps=1.2,min_samples=12)
db_labels=db.fit_predict(X_scaled[idx])
print("DBSCAN labels:")
print(pd.Series(db_labels).value_counts().sort_index())
```

```python
iso=IsolationForest(n_estimators=150,contamination=0.02,random_state=42,n_jobs=2)
iso_pred=iso.fit_predict(X_scaled)
df["IsolationForestAnomaly"]=(iso_pred==-1).astype(int)
print("Isolation Forest anomalies:",df["IsolationForestAnomaly"].sum())
```

```python
plt.scatter(X_scaled[idx,0],X_scaled[idx,1],c=db_labels,s=8,cmap="tab20")
plt.xlabel("Standardised CGPA")
plt.ylabel("Standardised Attendance")
plt.title("DBSCAN Clustering / Noise")
plt.show()
```
