```python
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV, validation_curve, learning_curve
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from scipy.stats import randint

target="PlacementStatus"; drops=["PlacementStatus","Salary Package","StudentID","IsAnomaly"]
X=df.drop(columns=drops,errors="ignore"); y=df[target].astype(int)
num=X.select_dtypes(include=np.number).columns.tolist(); cat=X.select_dtypes(include="object").columns.tolist()
pre=ColumnTransformer([("num",SimpleImputer(strategy="median"),num),
("cat",Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("onehot",OneHotEncoder(handle_unknown="ignore"))]),cat)])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
# Use a representative training sample for repeated search/curve calculations.
tune_idx=np.random.RandomState(42).choice(len(Xtr),size=min(10000,len(Xtr)),replace=False)
Xtr=Xtr.iloc[tune_idx]
ytr=ytr.iloc[tune_idx]
pipe=Pipeline([("preprocessor",pre),("model",DecisionTreeClassifier(random_state=42))])

grid={"model__max_depth":[3,5,8,12],"model__min_samples_split":[2,5]}
grid_search=GridSearchCV(pipe,grid,cv=2,scoring="accuracy",n_jobs=2)
grid_search.fit(Xtr,ytr)
print("GridSearch best parameters:",grid_search.best_params_)
print("GridSearch best score:",grid_search.best_score_)
```

```python
random_search=RandomizedSearchCV(pipe,
{"model__max_depth":randint(3,15),"model__min_samples_split":randint(2,12)},
n_iter=3,cv=2,scoring="accuracy",random_state=42,n_jobs=2)
random_search.fit(Xtr,ytr)
print("RandomSearch best parameters:",random_search.best_params_)
print("RandomSearch best score:",random_search.best_score_)
```

```python
train_sizes,train_scores,val_scores=learning_curve(
grid_search.best_estimator_,Xtr,ytr,cv=2,train_sizes=np.linspace(.2,1.0,5),
scoring="accuracy",n_jobs=2)
plt.plot(train_sizes,train_scores.mean(axis=1),marker="o",label="Training")
plt.plot(train_sizes,val_scores.mean(axis=1),marker="o",label="Validation")
plt.xlabel("Training examples"); plt.ylabel("Accuracy")
plt.title("Learning Curve"); plt.legend(); plt.show()
```

```python
param_range=[3,5,8,12]
train_scores,val_scores=validation_curve(
pipe,Xtr,ytr,param_name="model__max_depth",param_range=param_range,cv=2,
scoring="accuracy",n_jobs=2)
plt.plot(param_range,train_scores.mean(axis=1),marker="o",label="Training")
plt.plot(param_range,val_scores.mean(axis=1),marker="o",label="Validation")
plt.xlabel("Max Depth"); plt.ylabel("Accuracy")
plt.title("Validation Curve"); plt.legend(); plt.show()
```
