```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
import shap

target="PlacementStatus"
drop_cols=["PlacementStatus","Salary Package","StudentID","IsAnomaly"]
X=df.drop(columns=drop_cols,errors="ignore")
y=df[target].astype(int)
num_cols=X.select_dtypes(include=np.number).columns.tolist()
cat_cols=X.select_dtypes(include="object").columns.tolist()
pre=ColumnTransformer([
("num",SimpleImputer(strategy="median"),num_cols),
("cat",Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("onehot",OneHotEncoder(handle_unknown="ignore"))]),cat_cols)])
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
Xtr=pre.fit_transform(X_train); Xte=pre.transform(X_test)
print("Prepared training data:",Xtr.shape)
```

```python
xgb=XGBClassifier(n_estimators=120,max_depth=4,learning_rate=.08,subsample=.8,colsample_bytree=.8,eval_metric="logloss",random_state=42,n_jobs=2)
lgb=LGBMClassifier(n_estimators=120,max_depth=5,learning_rate=.08,verbosity=-1,random_state=42,n_jobs=2)
xgb.fit(Xtr,y_train); lgb.fit(Xtr,y_train)
xgb_pred=xgb.predict(Xte); lgb_pred=lgb.predict(Xte)
print("XGBoost Accuracy:",accuracy_score(y_test,xgb_pred))
print("LightGBM Accuracy:",accuracy_score(y_test,lgb_pred))
```

```python
scores=pd.DataFrame({"Model":["XGBoost","LightGBM"],
"Accuracy":[accuracy_score(y_test,xgb_pred),accuracy_score(y_test,lgb_pred)]})
display(scores)
plt.bar(scores["Model"],scores["Accuracy"])
plt.ylim(0,1); plt.ylabel("Accuracy"); plt.title("XGBoost vs LightGBM")
plt.show()
```

```python
sample_n=min(500,len(X_test))
X_sample=Xte[:sample_n]
explainer=shap.TreeExplainer(xgb)
shap_values=explainer.shap_values(X_sample)
print("SHAP explanation generated for",sample_n,"test records.")
shap.summary_plot(shap_values,X_sample,feature_names=pre.get_feature_names_out(),show=True,max_display=12)
```
