```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
import joblib

target="PlacementStatus"; drops=["PlacementStatus","Salary Package","StudentID","IsAnomaly"]
X=df.drop(columns=drops,errors="ignore"); y=df[target].astype(int)
num=X.select_dtypes(include=np.number).columns.tolist(); cat=X.select_dtypes(include="object").columns.tolist()
pre=ColumnTransformer([("num",SimpleImputer(strategy="median"),num),
("cat",Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("onehot",OneHotEncoder(handle_unknown="ignore"))]),cat)])
model=Pipeline([("preprocessor",pre),("model",RandomForestClassifier(n_estimators=80,max_depth=10,random_state=42,n_jobs=2))])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
model.fit(Xtr,ytr)
joblib.dump(model,"batch_model.joblib")
print("Model trained and saved.")
```

```python
# Batch scoring
batch=Xte.head(100).copy()
batch_predictions=model.predict(batch)
batch_output=batch.copy()
batch_output["PredictedPlacementStatus"]=batch_predictions
print("Batch scoring completed.")
display(batch_output.head())
```

```python
# Population Stability Index (PSI)
def psi(expected, actual, bins=10):
    edges=np.unique(np.quantile(expected,np.linspace(0,1,bins+1)))
    if len(edges)<3:
        return 0.0
    expected_counts,_=np.histogram(expected,bins=edges)
    actual_counts,_=np.histogram(actual,bins=edges)
    expected_pct=np.clip(expected_counts/len(expected),1e-6,None)
    actual_pct=np.clip(actual_counts/len(actual),1e-6,None)
    return float(np.sum((actual_pct-expected_pct)*np.log(actual_pct/expected_pct)))

reference=df["CGPA"].dropna().values
current=(df["CGPA"].dropna()+0.25).values
psi_value=psi(reference,current)
print("CGPA PSI:",round(psi_value,4))
```

```python
if psi_value < 0.10:
    status="No significant drift"
elif psi_value < 0.25:
    status="Moderate drift - monitor"
else:
    status="Significant drift - retraining recommended"
print("Drift status:",status)
```

```python
# Retraining decision
if psi_value >= 0.25:
    print("RETRAIN: PSI exceeded the 0.25 threshold.")
else:
    print("NO RETRAINING: PSI is below the retraining threshold.")
```
