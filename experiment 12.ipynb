```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
import shap

target="PlacementStatus"; drops=["PlacementStatus","Salary Package","StudentID","IsAnomaly"]
X=df.drop(columns=drops,errors="ignore"); y=df[target].astype(int)
num=X.select_dtypes(include=np.number).columns.tolist(); cat=X.select_dtypes(include="object").columns.tolist()
pre=ColumnTransformer([("num",SimpleImputer(strategy="median"),num),
("cat",Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("onehot",OneHotEncoder(handle_unknown="ignore"))]),cat)])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
Xtr=pre.fit_transform(Xtr); Xte=pre.transform(Xte)
xgb=XGBClassifier(n_estimators=100,max_depth=4,learning_rate=.08,eval_metric="logloss",random_state=42,n_jobs=2)
lgb=LGBMClassifier(n_estimators=100,max_depth=5,learning_rate=.08,verbosity=-1,random_state=42,n_jobs=2)
xgb.fit(Xtr,ytr); lgb.fit(Xtr,ytr)
print("XGBoost Accuracy:",accuracy_score(yte,xgb.predict(Xte)))
print("LightGBM Accuracy:",accuracy_score(yte,lgb.predict(Xte)))
```

```python
explainer=shap.TreeExplainer(xgb)
sv=explainer.shap_values(Xte[:500])
shap.summary_plot(sv,Xte[:500],feature_names=pre.get_feature_names_out(),max_display=12,show=True)
```

```python
importance=pd.Series(xgb.feature_importances_,index=pre.get_feature_names_out()).sort_values(ascending=False).head(12)
plt.figure(figsize=(8,5))
plt.barh(importance.index[::-1],importance.values[::-1])
plt.title("Top XGBoost Features")
plt.xlabel("Importance")
plt.show()
```
