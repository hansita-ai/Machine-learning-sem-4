```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
import joblib, os

target="PlacementStatus"; drops=["PlacementStatus","Salary Package","StudentID","IsAnomaly"]
X=df.drop(columns=drops,errors="ignore"); y=df[target].astype(int)
num=X.select_dtypes(include=np.number).columns.tolist(); cat=X.select_dtypes(include="object").columns.tolist()
pre=ColumnTransformer([("num",SimpleImputer(strategy="median"),num),
("cat",Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),("onehot",OneHotEncoder(handle_unknown="ignore"))]),cat)])
model=Pipeline([("preprocessor",pre),("model",RandomForestClassifier(n_estimators=100,max_depth=12,random_state=42,n_jobs=2))])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
model.fit(Xtr,ytr)
print("Modular pipeline training completed.")
```

```python
joblib.dump(model,"placement_model.joblib")
print("Saved model: placement_model.joblib")
print("Test accuracy:",model.score(Xte,yte))
```

```python
fastapi_code = '''
from fastapi import FastAPI
from pydantic import BaseModel
import joblib

app = FastAPI(title="Placement Prediction API")
model = joblib.load("placement_model.joblib")

class Student(BaseModel):
    CGPA: float
    AttendancePercent: float
    Internships: int
    Projects: int
    Certifications: int
    AptitudeTestScore: float
    SoftSkillsRating: float
    CodingTestScore: float
    MockInterviewScore: float

@app.get("/")
def home():
    return {"message": "Placement Prediction API is running"}

@app.post("/predict")
def predict(student: Student):
    data = student.model_dump()
    prediction = model.predict([data])
    return {"PlacementStatus": int(prediction[0])}
'''
open("main.py","w").write(fastapi_code)
print("Created FastAPI file: main.py")
print("Run in Colab/terminal with: uvicorn main:app --reload")
```
